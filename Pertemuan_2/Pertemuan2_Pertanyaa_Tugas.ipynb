```python
img = cv.imread("/content/drive/MyDrive/PCVK/KTM lama.jpg")

# BGR ke RGB
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Kanal R-B (Green dibuat 0)
rb = img_rgb.copy()
rb[:,:,1] = 0

# Kanal G-B (Red dibuat 0)
gb = img_rgb.copy()
gb[:,:,0] = 0

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(rb)
plt.title("Kombinasi Kanal Merah-Biru")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(gb)
plt.title("Kombinasi Kanal Hijau-Biru")
plt.axis("off")

plt.show()
```

**TUGAS PRAKTIKUM - Analisa**

Berapa rata-rata intensitas kecerahan (brightness) pada citra KTM lama?

```python
img = cv.imread("/content/drive/MyDrive/PCVK/KTM lama.jpg")

# Mengubah citra BGR menjadi grayscale
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# Menghitung rata-rata intensitas piksel
mean_intensity = np.mean(gray)

print("Ukuran citra:", gray.shape)
print("Rata-rata intensitas:", mean_intensity)

# Menampilkan citra grayscale
plt.imshow(gray, cmap="gray")
plt.title("Citra Grayscale KTM")
plt.axis("off")
plt.show()
```

**Kesimpulan**

Berdasarkan hasil perhitungan, citra KTM lama memiliki rata-rata intensitas sebesar 136.82. Nilai tersebut menunjukkan bahwa tingkat kecerahan citra berada pada tingkat sedang karena berada di sekitar nilai tengah rentang intensitas grayscale (0–255). Program berhasil menghitung karakteristik citra berdasarkan nilai piksel tanpa hanya mengandalkan pengamatan visual.

**TUGAS PRAKTIKUM - Aplikasi**

**Studi Kasus 1**

```python
img = cv.imread("/content/drive/MyDrive/PCVK/KTM lama.jpg")
# Mengambil ukuran citra
h, w = img.shape[:2]
# Membuat tinggi bar secara adaptif
bar_height = int(h * 0.12)
# Membuat canvas tambahan untuk bar identitas
canvas = np.zeros((h + bar_height, w, 3), dtype=np.uint8)
# Menempatkan gambar asli
canvas[0:h, 0:w] = img
# Warna bar identitas
canvas[h:h+bar_height, :] = (40, 40, 40)
# Menentukan ukuran font berdasarkan tinggi gambar
font = cv.FONT_HERSHEY_SIMPLEX
font_scale = h / 500
thickness = max(1, int(h / 300))
# Posisi teks adaptif
text_x = int(w * 0.05)
text_y = h + int(bar_height * 0.7)
# Menambahkan label identitas
cv.putText(
    canvas,
    "DOKUMENTASI KEGIATAN",
    (text_x, text_y),
    font,
    font_scale,
    (255,255,255),
    thickness,
    cv.LINE_AA
)
# Konversi warna untuk matplotlib
result = cv.cvtColor(canvas, cv.COLOR_BGR2RGB)
# Menampilkan hasil
plt.figure(figsize=(8,6))
plt.imshow(result)
plt.title("Foto dengan Label Identitas Adaptif")
plt.axis("off")
plt.show()
```

**Studi Kasus 2**

```python
img = cv.imread("/content/drive/MyDrive/PCVK/KTM lama.jpg")
# Ukuran asli gambar
h, w = img.shape[:2]
# Membuat canvas persegi mengikuti sisi terbesar gambar
size = max(h, w)
canvas = np.zeros((size, size, 3), dtype=np.uint8)
# Warna background canvas
canvas[:] = (255,255,255)
# Menghitung ukuran resize agar rasio tetap
scale = min(size/w, size/h)
new_w = int(w * scale)
new_h = int(h * scale)
# Resize gambar
img_resize = cv.resize(img, (new_w, new_h))
# Posisi agar gambar berada di tengah
x = (size - new_w)//2
y = (size - new_h)//2
# Tempel gambar ke canvas
canvas[y:y+new_h, x:x+new_w] = img_resize
# Membuat ukuran font adaptif berdasarkan ukuran gambar
font = cv.FONT_HERSHEY_SIMPLEX
font_scale = size / 500
thickness = max(1, int(size/300))
# Posisi logo adaptif
text_x = int(size * 0.05)
text_y = int(size * 0.90)
cv.putText(
    canvas,
    "MY KTM",
    (text_x,text_y),
    font,
    font_scale,
    (0,0,255),
    thickness,
    cv.LINE_AA
)
# Tampilkan hasil
canvas_rgb = cv.cvtColor(canvas, cv.COLOR_BGR2RGB)

plt.figure(figsize=(6,6))
plt.imshow(canvas_rgb)
plt.title("menampilkan foto produk dalam bingkai persegi 1:1")
plt.axis("off")
plt.show()
```
